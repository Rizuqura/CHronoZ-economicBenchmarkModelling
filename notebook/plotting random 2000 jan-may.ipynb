# Plotting Final — Economic Series Movement Around May 2000

Goal: inspect **how all economic series move during the same calendar year as the reference event, May 2000**. Each series is shown in a shared event window from January 2000 through December 2000, with May 2000 marked clearly. The method is applied consistently to every CSV in `data/cleaned`.

For each series, the notebook plots the observed values during 2000 and the 2- and 4-observation rolling means, plus a standardized view using that series' full-history mean and standard deviation for context. Series with no observations in 2000 are reported as unavailable for that year's movement view rather than silently substituted. Observation frequency may differ between series, so 2 and 4 rolling observations do not always equal 2 and 4 months.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# ---- Shared observation-year controls ----
TARGET_YEAR = 2000
TARGET_MONTH = 5
ROLLING_WINDOWS = (2, 4)

# Event window: the whole calendar year containing the reference month.
WINDOW_START = pd.Timestamp(TARGET_YEAR, 1, 1)
WINDOW_END = pd.Timestamp(TARGET_YEAR, 5, 31)
TARGET_START = pd.Timestamp(TARGET_YEAR, TARGET_MONTH, 1)
TARGET_END = TARGET_START + pd.offsets.MonthEnd(1)

# Locate data regardless of whether Jupyter starts in repository root or notebook folder.
candidates = [Path('data/cleaned'), Path('../data/cleaned'), Path('../../data/cleaned')]
cleaned_dir = next((p.resolve() for p in candidates if p.exists()), None)
if cleaned_dir is None:
    raise FileNotFoundError('Cannot find data/cleaned. Start Jupyter from repository root or notebook folder.')

csv_paths = sorted(cleaned_dir.glob('*.csv'))
if not csv_paths:
    raise FileNotFoundError(f'No CSV files found in {cleaned_dir}')

print(f'Common observation year: {TARGET_YEAR}')
print(f'Event month highlighted: {TARGET_START:%B %Y}')
print(f'Window shown: {WINDOW_START:%Y-%m-%d} to {WINDOW_END:%Y-%m-%d}')
print(f'Rolling windows for every series: {ROLLING_WINDOWS} observations')
print(f'CSV files discovered: {len(csv_paths)}')

series_by_id = {}
load_issues = []
for csv_path in csv_paths:
    series_id = csv_path.stem
    try:
        raw = pd.read_csv(csv_path)
        if not {'date', 'value'}.issubset(raw.columns):
            load_issues.append((series_id, f"Missing date/value columns: {list(raw.columns)}"))
            continue
        raw['date'] = pd.to_datetime(raw['date'], errors='coerce')
        raw['value'] = pd.to_numeric(raw['value'], errors='coerce')
        s = (raw.dropna(subset=['date', 'value'])
                 .sort_values('date')
                 .drop_duplicates('date', keep='last')
                 .set_index('date')['value'])
        if s.empty:
            load_issues.append((series_id, 'No valid date/value observations'))
            continue
        series_by_id[series_id] = s
    except Exception as exc:
        load_issues.append((series_id, str(exc)))

if not series_by_id:
    raise ValueError('No valid economic series could be loaded.')

print(f'Valid series loaded: {len(series_by_id)}')
if load_issues:
    print(f'Series skipped due to data/schema issues: {len(load_issues)}')
    display(pd.DataFrame(load_issues, columns=['series_id', 'issue']))

# Summary explicitly checks data availability during the target year and month.
summary_rows = []
for series_id, s in sorted(series_by_id.items()):
    year_values = s.loc[(s.index >= WINDOW_START) & (s.index <= WINDOW_END)]
    may_values = s.loc[(s.index >= TARGET_START) & (s.index <= TARGET_END)]
    mu, sigma = s.mean(), s.std(ddof=1)
    summary_rows.append({
        'series_id': series_id,
        'first_date': s.index.min(),
        'last_date': s.index.max(),
        'observations_in_2000': len(year_values),
        'observations_in_may_2000': len(may_values),
        'may_2000_value_first_obs': may_values.iloc[0] if len(may_values) else np.nan,
        '2000_value_first_obs': year_values.iloc[0] if len(year_values) else np.nan,
        '2000_value_last_obs': year_values.iloc[-1] if len(year_values) else np.nan,
        'full_history_mean': mu,
        'full_history_sample_std': sigma
    })

summary = pd.DataFrame(summary_rows).sort_values('series_id').reset_index(drop=True)
display(summary)
```

## All-series movement during calendar year 2000

Each chart is zoomed to January–December 2000 so you can inspect the movement through that year, rather than looking at the entire multi-decade history. May is highlighted as the shared reference month. The standardized panel provides historical context, while the raw panel preserves the series' original units.

```python
# Apply the same 2000 event-window view to every economic series.
for series_id, s in sorted(series_by_id.items()):
    year_s = s.loc[(s.index >= WINDOW_START) & (s.index <= WINDOW_END)]
    if year_s.empty:
        print(f'{series_id}: no observations in {TARGET_YEAR}; cannot plot its movement during that year.')
        continue

    full_mean = s.mean()
    full_std = s.std(ddof=1)
    year_z = ((year_s - full_mean) / full_std) if pd.notna(full_std) and full_std != 0 else None
    rolling_2 = year_s.rolling(window=ROLLING_WINDOWS[0], min_periods=1).mean()
    rolling_4 = year_s.rolling(window=ROLLING_WINDOWS[1], min_periods=1).mean()
    may_mask = (year_s.index >= TARGET_START) & (year_s.index <= TARGET_END)

    fig, axes = plt.subplots(2, 1, figsize=(13, 7), sharex=True, constrained_layout=True)
    for ax in axes:
        ax.axvspan(TARGET_START, TARGET_END, color='orange', alpha=0.22,
                   label=f'Reference month: {TARGET_START:%B %Y}')

    if year_z is not None:
        axes[0].plot(year_z.index, year_z.values, marker='o', markersize=3,
                     linewidth=1.2, label='Z-score (full-history baseline)')
        axes[0].axhline(0, color='black', linewidth=1, label='Full-history mean')
        axes[0].axhline(1, linestyle='--', alpha=0.5, label='+/- 1 std')
        axes[0].axhline(-1, linestyle='--', alpha=0.5)
        if may_mask.any():
            axes[0].scatter(year_z.loc[may_mask].index, year_z.loc[may_mask].values,
                            color='darkorange', edgecolor='black', zorder=5, label='May observation(s)')
        axes[0].set_ylabel('Z-score')
        axes[0].legend(fontsize=8, ncol=2)
    else:
        axes[0].text(0.5, 0.5, 'Z-score unavailable: insufficient variation',
                     transform=axes[0].transAxes, ha='center', va='center')
        axes[0].set_ylabel('Z-score')

    axes[0].set_title(f'{series_id} — movement during {TARGET_YEAR} (standardized)')
    axes[0].grid(True, alpha=0.25)

    axes[1].plot(year_s.index, year_s.values, marker='o', markersize=3,
                 linewidth=1.2, label='Observed value')
    axes[1].plot(year_s.index, rolling_2.values, linewidth=1.2,
                 label='Rolling mean (2 observations)')
    axes[1].plot(year_s.index, rolling_4.values, linewidth=1.5,
                 label='Rolling mean (4 observations)')
    if may_mask.any():
        axes[1].scatter(year_s.loc[may_mask].index, year_s.loc[may_mask].values,
                        color='darkorange', edgecolor='black', zorder=5, label='May observation(s)')
    axes[1].set_title(f'Raw {series_id} values during {TARGET_YEAR}')
    axes[1].set_ylabel('Original units')
    axes[1].set_xlabel('Date')
    axes[1].legend(fontsize=8, ncol=2)
    axes[1].grid(True, alpha=0.25)
    plt.show()

    print(f'{series_id}: {len(year_s)} observation(s) in {TARGET_YEAR}; '
          f'{int(may_mask.sum())} observation(s) in May.')
    display(year_s.rename('value').to_frame())
```

## Interpretation checklist

- Follow each series from the start to the end of 2000 and note whether it rises, falls, turns, or remains relatively stable.
- Compare movements around May 2000 across series using the standardized panel; use the raw panel for each series' original units.
- Confirm whether each series actually has a May observation. Monthly, quarterly, and irregular series may show different numbers of points.
- Rolling means are calculated from observations within the displayed 2000 window; the first few values therefore use fewer observations when `min_periods=1`.
- The standardized baseline uses the full available history, including data after 2000. This is suitable for retrospective visualization, not a no-look-ahead trading test.
- Similar-looking movements do not prove correlation, causality, or mean reversion; those require separate statistical tests.

## Benchmark-centered spread and 4-observation ROC-weighted spread

The benchmark comes from the top summary table: each series' full-history mean and sample standard deviation. Centering uses `X_centered = X - full_history_mean`, so the historical mean maps to zero. The final spread is the last available observation in January–May 2000 expressed relative to that benchmark and scaled by the series' standard deviation: `spread_z = (last_value - full_history_mean) / full_history_sample_std`. Thus, spread is measured in standard-deviation units and is comparable across series. The second metric multiplies this benchmark-relative spread by the 4-observation backward ROC, `(X_t / X_(t-4)) - 1`, evaluated at the horizon endpoint.

Series with missing/zero standard deviation or undefined ROC are shown as unavailable in the raw table; for the exploratory polynomial plots only, unavailable fit inputs are filled with 0 and clearly labeled. The benchmark mean/std use each series' full history, including observations after 2000, so this is retrospective analysis rather than a no-look-ahead forecast.

```python
# Benchmark-relative compact table: use full-history mean and sample std from the top summary table.
# Spread is the final observation's standardized distance from the mean benchmark (mean -> 0).
# The 4-observation backward ROC is applied later in the polynomial section.

DERIVATIVE_START = pd.Timestamp(2000, 1, 1)
DERIVATIVE_END = pd.Timestamp(2000, 5, 31)

compact_rows = []

for series_id, s in sorted(series_by_id.items()):
    horizon = s.loc[(s.index >= DERIVATIVE_START) & (s.index <= DERIVATIVE_END)].copy()
    full_mean = s.mean()
    full_std = s.std(ddof=1)

    if horizon.empty:
        compact_rows.append({
            'data_name': series_id,
            'benchmark_mean_full_history': full_mean,
            'benchmark_std_full_history': full_std,
            'last_data_date': pd.NaT,
            'last_data_value': np.nan,
            'observations_in_horizon': 0,
            'centered_last_value': np.nan,
            'spread_zscore_from_zero_benchmark': np.nan,
            'status': 'No observations in Jan-May 2000'
        })
        continue

    last_date = horizon.index[-1]
    last_value = float(horizon.iloc[-1])
    centered_last = last_value - full_mean
    spread_z = centered_last / full_std if pd.notna(full_std) and full_std != 0 else np.nan
    status = 'OK' if pd.notna(spread_z) else 'Spread unavailable: std is zero or undefined'

    compact_rows.append({
        'data_name': series_id,
        'benchmark_mean_full_history': full_mean,
        'benchmark_std_full_history': full_std,
        'last_data_date': last_date,
        'last_data_value': last_value,
        'observations_in_horizon': len(horizon),
        'centered_last_value': centered_last,
        'spread_zscore_from_zero_benchmark': spread_z,
        'status': status
    })

compact_derivative_spread = pd.DataFrame(compact_rows).sort_values('data_name').reset_index(drop=True)
display(compact_derivative_spread)
```

## Polynomial fit: benchmark-relative spread and spread × 4-observation backward ROC

The spread is the last observation's standardized distance from the full-history mean benchmark: `(X_t - mean_full_history) / std_full_history`. Since the series is centered, the mean benchmark is zero; dividing by standard deviation expresses distance in standard-deviation units. The second metric multiplies this spread by the original series' 4-observation backward ROC, `(X_t / X_(t-4)) - 1`, measured at the last observation in the January–May 2000 horizon.

The x-axis is the alphabetical series index, so these are cross-series comparison curves, not time-series fits. ROC is a relative change, not a calculus derivative; the 4-observation lag spans different calendar durations across series. For exploratory polynomial fitting only, unavailable values are filled with 0.

```python
# Compare benchmark-relative standardized spread with spread weighted by 4-observation backward ROC.
# Missing values are preserved in the displayed table; NaN fit inputs are set to 0 for exploratory plots.

POLYNOMIAL_DEGREE = 2
ROC_LAG_OBSERVATIONS = 4

poly_data = compact_derivative_spread.copy().sort_values('data_name').reset_index(drop=True)
poly_data['spread_for_fit'] = pd.to_numeric(
    poly_data['spread_zscore_from_zero_benchmark'], errors='coerce'
)

roc_values = []
roc_dates = []
roc_lag_dates = []
for series_id in poly_data['data_name']:
    s = series_by_id[series_id].sort_index()
    horizon = s.loc[(s.index >= DERIVATIVE_START) & (s.index <= DERIVATIVE_END)]
    if horizon.empty:
        roc_values.append(np.nan)
        roc_dates.append(pd.NaT)
        roc_lag_dates.append(pd.NaT)
        continue

    end_date = horizon.index[-1]
    end_position = s.index.get_indexer([end_date])[0]
    lag_position = end_position - ROC_LAG_OBSERVATIONS
    if lag_position < 0:
        roc_values.append(np.nan)
        roc_dates.append(end_date)
        roc_lag_dates.append(pd.NaT)
        continue

    x_now = float(s.iloc[end_position])
    x_lag = float(s.iloc[lag_position])
    if x_lag == 0 or not np.isfinite(x_lag):
        roc_values.append(np.nan)
    else:
        roc_values.append((x_now / x_lag) - 1.0)
    roc_dates.append(end_date)
    roc_lag_dates.append(s.index[lag_position])

poly_data['roc_4_observation'] = roc_values
poly_data['roc_end_date'] = roc_dates
poly_data['roc_lag_date'] = roc_lag_dates
poly_data['roc_for_fit'] = pd.to_numeric(poly_data['roc_4_observation'], errors='coerce')
poly_data['spread_times_roc_4'] = poly_data['spread_for_fit'] * poly_data['roc_for_fit']

# Fill only the polynomial fitting vectors; keep NaNs in the audit/display columns.
poly_data['spread_for_fit_plot'] = poly_data['spread_for_fit'].fillna(0.0)
poly_data['spread_times_roc_4_plot'] = poly_data['spread_times_roc_4'].fillna(0.0)
poly_data['series_index'] = np.arange(len(poly_data), dtype=float)

if len(poly_data) <= POLYNOMIAL_DEGREE:
    raise ValueError(f'Need more than {POLYNOMIAL_DEGREE} series to fit degree {POLYNOMIAL_DEGREE}.')

x_smooth = np.linspace(poly_data['series_index'].min(), poly_data['series_index'].max(), 500)
spread_coefficients = np.polyfit(poly_data['series_index'], poly_data['spread_for_fit_plot'], POLYNOMIAL_DEGREE)
spread_polynomial = np.poly1d(spread_coefficients)

plt.figure(figsize=(14, 6))
plt.scatter(poly_data['series_index'], poly_data['spread_for_fit_plot'], alpha=0.75,
            label='Benchmark-relative spread in std units (NaN → 0 for fit)')
plt.plot(x_smooth, spread_polynomial(x_smooth), linewidth=2,
         label=f'Polynomial fit, degree {POLYNOMIAL_DEGREE}')
plt.axhline(0, color='black', linewidth=1, alpha=0.7, label='Mean benchmark = 0')
plt.title('Polynomial fit of benchmark-relative spread — January–May 2000')
plt.xlabel('Economic series index (alphabetical order)')
plt.ylabel('Spread (standard deviations from benchmark)')
plt.grid(True, alpha=0.25)
plt.legend()
plt.show()

weighted_coefficients = np.polyfit(poly_data['series_index'], poly_data['spread_times_roc_4_plot'], POLYNOMIAL_DEGREE)
weighted_polynomial = np.poly1d(weighted_coefficients)
plt.figure(figsize=(14, 6))
plt.scatter(poly_data['series_index'], poly_data['spread_times_roc_4_plot'], alpha=0.75,
            label='Spread × 4-observation ROC (NaN → 0 for fit)')
plt.plot(x_smooth, weighted_polynomial(x_smooth), linewidth=2,
         label=f'Polynomial fit, degree {POLYNOMIAL_DEGREE}')
plt.axhline(0, color='black', linewidth=1, alpha=0.7, label='Mean benchmark = 0')
plt.title('Polynomial fit of benchmark-relative spread × 4-observation ROC — January–May 2000')
plt.xlabel('Economic series index (alphabetical order)')
plt.ylabel('Spread (std units) × ROC')
plt.grid(True, alpha=0.25)
plt.legend()
plt.show()

print('Series included in polynomial plots; spread is benchmark-relative and ROC is measured at the horizon endpoint:')
display(poly_data[[
    'data_name', 'series_index', 'benchmark_mean_full_history', 'benchmark_std_full_history',
    'last_data_date', 'last_data_value', 'centered_last_value',
    'spread_zscore_from_zero_benchmark', 'roc_lag_date', 'roc_end_date',
    'roc_4_observation', 'spread_times_roc_4'
]].merge(
    compact_derivative_spread[['data_name', 'benchmark_mean_full_history', 'benchmark_std_full_history']],
    on='data_name', how='left', suffixes=('', '_table')
).drop(columns=['benchmark_mean_full_history_table', 'benchmark_std_full_history_table']))
```

## Category-level aggregation and polynomial comparisonThis final section uses the explicit category registry in `dataCleansing.py` / README, rather than inferring categories from series names. Each member series is standardized using its own full-history mean and sample standard deviation. To make weekly, monthly, and quarterly series comparable on a common timeline, standardized observations are converted to calendar-month values using the last observation available in each month; the category output for a month is the cross-sectional mean of available member-series z-scores. A category value is therefore a composite standardized state, not a raw-unit economic index.For the January–May 2000 window, the category endpoint is its last available monthly composite. Its benchmark-relative spread is the endpoint minus the category composite's full-history mean, divided by the category composite's full-history sample standard deviation. The category's 4-observation ROC is calculated from its monthly composite series (four calendar-month observations back), and the interaction score is spread × ROC. The category outputs are then compared with a degree-2 polynomial across the six categories. Since category index is categorical, the polynomial is only a descriptive cross-category curve—not a temporal cycle model. Full-history normalization includes future data relative to 2000, so this is retrospective exploration, not a no-look-ahead test.

```python
# Category aggregation using the explicit registry in dataCleansing.py.
# Keep this as an appended exploratory section; earlier cells are unchanged.

CATEGORY_MEMBERS = {
    'credit': ['BUSLOANS', 'TOTALSL', 'DRTSCILM', 'DRSDCILM', 'DRBLACBS'],
    'external_structural': ['EXPGSC1', 'IMPGSC1', 'OPHNFB', 'TCU'],
    'growth_activity': ['INDPRO', 'IPMAN', 'PCEC96', 'W875RX1', 'RPI', 'RRSFS', 'HOUST', 'PERMIT'],
    'inflation': ['CPIAUCSL', 'CPILFESL', 'PCEPI', 'PCEPILFE', 'PPIFID', 'PPIACO'],
    'labor': ['PAYEMS', 'UNRATE', 'ICSA', 'AWHAETP', 'AWHMAN', 'JTSJOL', 'CIVPART', 'CES0500000003'],
    'policy_liquidity': ['FEDFUNDS', 'TOTRESNS', 'WRESBAL', 'WALCL', 'M2SL'],
}

CATEGORY_WINDOW_START = pd.Timestamp('2000-01-01')
CATEGORY_WINDOW_END = pd.Timestamp('2000-05-31')
category_member_audit = []
standardized_monthly = {}

# Standardize each series first, then map observations to calendar months.
for category, members in CATEGORY_MEMBERS.items():
    for series_id in members:
        if series_id not in series_by_id:
            category_member_audit.append({
                'category': category, 'series_id': series_id,
                'status': 'Series not loaded in series_by_id'
            })
            continue
        s = series_by_id[series_id].dropna().sort_index()
        mu = s.mean()
        sigma = s.std(ddof=1)
        if pd.isna(sigma) or sigma == 0:
            category_member_audit.append({
                'category': category, 'series_id': series_id,
                'status': 'Skipped: standard deviation zero/undefined'
            })
            continue
        z = (s - mu) / sigma
        # Monthly value is the last observed z-score in each calendar month.
        z.index = z.index.to_period('M')
        z_monthly = z.groupby(level=0).last()
        standardized_monthly[(category, series_id)] = z_monthly
        category_member_audit.append({
            'category': category, 'series_id': series_id,
            'status': 'Included', 'observations': len(s),
            'first_date': s.index.min(), 'last_date': s.index.max()
        })

# Build one category-level composite time series per category.
category_series = {}
for category, members in CATEGORY_MEMBERS.items():
    available = {
        sid: standardized_monthly[(category, sid)]
        for sid in members if (category, sid) in standardized_monthly
    }
    if not available:
        continue
    member_frame = pd.concat(available, axis=1).sort_index()
    # Average available member z-scores each month; do not fill missing member data with zero.
    category_series[category] = member_frame.mean(axis=1, skipna=True).dropna()

display(pd.DataFrame(category_member_audit))
if not category_series:
    raise ValueError('No category composites could be created; check data/cleaned and registry IDs.')

category_rows = []
category_output_series = {}
for category, composite in category_series.items():
    composite.index = composite.index.to_timestamp(how='end').normalize()
    full_mean = composite.mean()
    full_std = composite.std(ddof=1)
    horizon = composite.loc[
        (composite.index >= CATEGORY_WINDOW_START) &
        (composite.index <= CATEGORY_WINDOW_END)
    ]
    if horizon.empty:
        category_rows.append({
            'category': category, 'member_series_loaded': sum((cat, sid) in standardized_monthly for sid in CATEGORY_MEMBERS[category]),
            'category_endpoint_date': pd.NaT, 'category_output': np.nan,
            'category_mean': full_mean, 'category_std': full_std,
            'spread_zscore': np.nan, 'roc_4_calendar_months': np.nan,
            'spread_times_roc': np.nan, 'status': 'No category composite observations in Jan-May 2000'
        })
        continue

    end_date = horizon.index[-1]
    endpoint = float(horizon.iloc[-1])
    spread_z = (endpoint - full_mean) / full_std if pd.notna(full_std) and full_std != 0 else np.nan

    # The composite is monthly; locate the endpoint's four-calendar-month lag.
    endpoint_period = end_date.to_period('M')
    lag_period = endpoint_period - 4
    lag_timestamp = lag_period.to_timestamp(how='end').normalize()
    lag_value = composite.loc[composite.index.to_period('M') == lag_period]
    if len(lag_value) and float(lag_value.iloc[-1]) != 0:
        roc4 = endpoint / float(lag_value.iloc[-1]) - 1
    else:
        roc4 = np.nan
    interaction = spread_z * roc4 if pd.notna(spread_z) and pd.notna(roc4) else np.nan
    category_output_series[category] = composite
    category_rows.append({
        'category': category,
        'member_series_loaded': sum((category, sid) in standardized_monthly for sid in CATEGORY_MEMBERS[category]),
        'category_endpoint_date': end_date,
        'category_output': endpoint,
        'category_mean': full_mean,
        'category_std': full_std,
        'spread_zscore': spread_z,
        'roc_4_calendar_months': roc4,
        'spread_times_roc': interaction,
        'status': 'OK' if pd.notna(interaction) else 'Spread/ROC unavailable'
    })

category_summary = pd.DataFrame(category_rows).sort_values('category').reset_index(drop=True)
display(category_summary)

# Plot the category composite time series so the aggregated functions can be inspected.
for category, composite in category_output_series.items():
    plt.figure(figsize=(12, 4))
    plt.plot(composite.index, composite.values, marker='o', markersize=3, linewidth=1.2)
    plt.axhline(composite.mean(), color='black', linewidth=1, alpha=0.7, label='Category composite full-history mean')
    plt.axvspan(CATEGORY_WINDOW_START, CATEGORY_WINDOW_END, alpha=0.15, label='Jan-May 2000 evaluation window')
    plt.title(f'{category}: monthly category composite (mean member z-score)')
    plt.xlabel('Month')
    plt.ylabel('Mean member z-score')
    plt.grid(True, alpha=0.25)
    plt.legend()
    plt.show()

# Polynomial is applied after aggregation: one point per category.
category_poly = category_summary.dropna(subset=['spread_zscore', 'spread_times_roc']).copy()
category_poly = category_poly.sort_values('category').reset_index(drop=True)
if len(category_poly) > 2:
    category_poly['category_index'] = np.arange(len(category_poly), dtype=float)
    x_dense = np.linspace(category_poly['category_index'].min(), category_poly['category_index'].max(), 300)
    for metric, label, ylabel in [
        ('spread_zscore', 'Category-level benchmark-relative spread', 'Spread (category std units)'),
        ('spread_times_roc', 'Category-level spread × 4-month ROC', 'Spread × ROC')
    ]:
        coefficients = np.polyfit(category_poly['category_index'], category_poly[metric], deg=2)
        fitted = np.poly1d(coefficients)
        plt.figure(figsize=(11, 5))
        plt.scatter(category_poly['category_index'], category_poly[metric], s=65, label='Category output')
        plt.plot(x_dense, fitted(x_dense), linewidth=2, label='Degree-2 polynomial fit')
        plt.axhline(0, color='black', linewidth=1, alpha=0.7)
        plt.xticks(category_poly['category_index'], category_poly['category'], rotation=25, ha='right')
        plt.title(label + ' — January–May 2000 endpoint')
        plt.xlabel('Economic category')
        plt.ylabel(ylabel)
        plt.grid(True, alpha=0.25)
        plt.legend()
        plt.tight_layout()
        plt.show()
    print('Category-level polynomial coefficients (highest degree first):')
    display(pd.DataFrame([
        {'metric': metric, 'degree_2': coeff[0], 'degree_1': coeff[1], 'intercept': coeff[2]}
        for metric, coeff in [
            ('spread_zscore', np.polyfit(category_poly['category_index'], category_poly['spread_zscore'], 2)),
            ('spread_times_roc', np.polyfit(category_poly['category_index'], category_poly['spread_times_roc'], 2))
        ]
    ]))
else:
    print('Polynomial fitting skipped: fewer than three categories have valid endpoint scores.')
```
